```python
import networkx as nx
import matplotlib.pyplot as plt

# Create a graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

# Depth Limited Search
def dls(graph, node, limit, visited):
    visited.append(node)

    if limit == 0:
        return

    for neighbor in graph[node]:
        if neighbor not in visited:
            dls(graph, neighbor, limit - 1, visited)


# Iterative Deepening DFS
def iddfs(graph, start, max_depth):
    for depth in range(max_depth + 1):
        visited = []
        dls(graph, start, depth, visited)

        print("Depth", depth, ":", visited)

    return visited


# Perform IDDFS
start_node = 'A'
max_depth = 2

result = iddfs(graph, start_node, max_depth)

print("IDDFS Final Traversal:", result)


# Graph Visualization
G = nx.DiGraph()

for node, neighbors in graph.items():
    for neighbor in neighbors:
        G.add_edge(node, neighbor)

pos = {
    'A': (0, 2),
    'B': (-1, 1),
    'C': (1, 1),
    'D': (-1.5, 0),
    'E': (-0.5, 0),
    'F': (0.5, 0),
    'G': (1.5, 0)
}

plt.figure(figsize=(9, 6))

nx.draw(
    G, pos,
    with_labels=True,
    node_color='lightyellow',
    node_size=2000,
    font_size=14,
    font_weight='bold',
    arrows=True,
    arrowsize=20
)

plt.title("IDDFS Graph Visualization")
plt.show()
```
